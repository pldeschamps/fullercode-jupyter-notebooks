## First piece of code: NORWAY longitude


```python
#This piece of code calculates "NORWAY" longitude with more than 16 digits

import math

# Given data
lat_china = math.radians(39.1)
lon_china = math.radians(122.3)
lat_norway = math.radians(64.7)

# Desired angle between them
theta1 = math.acos(1 / math.sqrt(5))
theta2 = math.atan(2)

#checking both methods to calculate the edge of a regular icosahedron are similar
print(f"theta1 = {theta1:.20f}")
print(f"theta2 = {theta2:.20f}")

# Compute cos(delta_longitude)
cos_dlon = (math.cos(theta1) - math.sin(lat_china) * math.sin(lat_norway)) / (math.cos(lat_china) * math.cos(lat_norway))

# Clamp to [-1, 1] to avoid floating errors
cos_dlon = max(-1, min(1, cos_dlon))

# Two possible longitudes
delta_lon = math.acos(cos_dlon)
lon_norway_1 = lon_china + delta_lon
lon_norway_2 = lon_china - delta_lon

# Convert to degrees (normalize between -180,180)
def normalize(deg):
    deg = math.degrees(deg)
    while deg > 180: deg -= 360
    while deg < -180: deg += 360
    return deg

lon1 = normalize(lon_norway_1)
lon2 = normalize(lon_norway_2)

print(f"Possible NORWAY longitudes:")
print(f"  λ₁ = {lon1:.20f}°")
print(f"  λ₂ = {lon2:.20f}°")
```

## Second piece of code: calculate the coordinates of all the vertices of the North hemisphere with 16 digits.

```python
import math
import numpy as np
import json

# --- Array of verctors of fuller icosahedron vertices ---
vectors = []

# --- add 13 vectors ---
vectors.extend([None] * 13)  # Placeholder for 13 vectors

#vectors[0]= np.array([0.0, 0.0, 1], dtype=np.float64)  # North pole, for reference

# Original data

lat_china = math.radians(39.1)
lon_china = math.radians(122.3)
lat_norway = math.radians(64.7)

# length of the icosahedron edge (radians)
theta2 = math.atan(2)

# Calculate Norway longitude with 16 digits
cos_dlon = (math.cos(theta2) - math.sin(lat_china) * math.sin(lat_norway)) / (math.cos(lat_china) * math.cos(lat_norway))
lon_norway_2 = lon_china - delta_lon

print(f"lon_norway_2 = {lon_norway_2*180/math.pi:.20f}")

x_china = math.cos(lat_china) * math.cos(lon_china)
y_china = math.cos(lat_china) * math.sin(lon_china)
z_china = math.sin(lat_china)

print(f"x_china = {x_china:.20f}")
print(f"y_china = {y_china:.20f}")
print(f"z_china = {z_china:.20f}")

x_norway = math.cos(lat_norway) * math.cos(lon_norway_2)
y_norway = math.cos(lat_norway) * math.sin(lon_norway_2)
z_norway = math.sin(lat_norway)

print(f"x_norway = {x_norway:.20f}")
print(f"y_norway = {y_norway:.20f}")
print(f"z_norway = {z_norway:.20f}")

# --- Create vectors ---

vectors[1]= np.array([x_china, y_china, z_china], dtype=np.float64)  # vertice No. 1 (CHINA)
vectors[2]= np.array([x_norway, y_norway, z_china], dtype=np.float64)   # vertice No. 2 (NORWAY)

# --- function of normalization ---
def normalize(v):
    norm = np.linalg.norm(v)
    if norm == 0:
        return v
    return v / norm

# --- Rotaion function (Rodrigues Formula) ---
def rodrigues_rotate(v, k, theta):
    v = np.array(v, dtype=np.float64)
    k = np.array(k, dtype=np.float64)
    return v * math.cos(theta) + np.cross(k, v) * math.sin(theta) + k * (np.dot(k, v)) * (1 - math.cos(theta))

# --- Dictionary of vectors to make a json file ---
vertices = []
def append_vertex(id, vec):
    vertices.append({"id": id, "x": float(vec[0]), "y": float(vec[1]), "z": float(vec[2])})
# Vecteur 1
append_vertex(1, vectors[1])
# Vecteur 2
append_vertex(2, vectors[2])
#vertices.append({"id": 1, "x": float(vectors[1][0]), "y": float(vectors[1][1]), "z": float(vectors[1][2])})

# Vecteur 2
#vertices.append({"id": 2, "x": float(v2[0]), "y": float(v2[1]), "z": float(v2[2])})

# Vertices 3 to 6 by rotation of v1 around v2
labels = {1: 6, 2: 5, 3: 4, 4: 3}
# --- axis normalization ---
axis = normalize(vectors[2])
for m in [1, 2, 3, 4]:
    theta = 2 * math.pi * m / 5.0
    v_rot = rodrigues_rotate(vectors[1], axis, theta)
    vectors[labels[m]] = v_rot
    append_vertex(labels[m], v_rot)
    #vertices.append({"id": labels[m], "x": float(v_rot[0]), "y": float(v_rot[1]), "z": float(v_rot[2])})

# Vertices 7 to 12 by central symetry of vertices 1 to 6
for i in range(1, 7):
    v_sym = -vectors[i]
    vectors[i + 6] = v_sym
    append_vertex(i + 6, v_sym)

# --- final dictionnary ---
data = {"vertices": vertices}

# --- Écriture dans un fichier JSON ---
with open("vertices3.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print("✅ 'vertices.json' files created !")
```
